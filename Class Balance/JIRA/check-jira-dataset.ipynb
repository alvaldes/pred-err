```python
import pandas as pd
import os

# Directorio donde están los archivos CSV
directorio = './jira'

# Lista para almacenar los resultados
resultados = []

# Iterar sobre todos los archivos CSV en el directorio
for archivo in os.listdir(directorio):
    if archivo.endswith('.csv'):
        ruta_completa = os.path.join(directorio, archivo)
        
        # Leer el CSV
        df = pd.read_csv(ruta_completa)
        
        # Calcular los parámetros
        num_filas = len(df)
        num_atributos = len(df.columns)
        
        # Calcular LOC total (solo si existe la columna)
        loc = df['CountLineCode'].sum() if 'CountLineCode' in df.columns else 'N/A'

        # Calcular Defective Rate % para HeuBug y RealBug
        heu_defective_rate = 'N/A'
        real_defective_rate = 'N/A'
        
        if 'HeuBug' in df.columns:
            defective_rate_heubug = f"{round((df['HeuBug'].sum() / num_filas) * 100, 2)}%" if num_filas > 0 else "0%"
        
        if 'RealBug' in df.columns:
            defective_rate_realbug = f"{round((df['RealBug'].sum() / num_filas) * 100, 2)}%" if num_filas > 0 else "0%"
        # Guardar resultados
        resultados.append({
            'Archivo': archivo,
            'LOC total': loc,
            'Cantidad de Atributos': num_atributos,
            'Número de Filas': num_filas,
            'Defective rate (HueBug %)': defective_rate_heubug,
            'Defective rate (RealBug %)': defective_rate_realbug
        })

# Crear DataFrame con los resultados
df_resultados = pd.DataFrame(resultados)
df_resultados.sort_values(by='Archivo', inplace=True)
df_resultados
```
