```python
class_col = "City"
null_col = ["AverageTemperature","AverageTemperatureUncertainty"]
df_no_null = df[~df.isnull().any(axis=1)]
df_no_null
```

```python
df.groupby(class_col)[null_col].median()
```

```python
median_avgTemp = df.groupby(class_col)[null_col].transform("median")
median_avgTemp
```

```python
new_df = df[null_col].fillna(median_avgTemp)
```

```python
new_df.tail()
```

```python
new_df.isnull().sum()
```

```python
export_df = df.copy()

export_df[null_col] = new_df

export_df
```

```python
export_df.to_csv("inc-a_GlobalLandTemperaturesByMajorCity.csv")
```
