```python
from sklearn.datasets import load_wine

wine = load_wine(as_frame=True)
df = wine.data
df.head()
```

```python
import matplotlib.pyplot as plt

df.boxplot()
plt.show()
```

```python
from sklearn.preprocessing import StandardScaler
import pandas as pd

scaler = StandardScaler()
X_scaled = scaler.fit_transform(df)

df2 = pd.DataFrame(X_scaled)
df2.boxplot()
plt.show()
```

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=3) # o 0.85 para umbral de varianza acumulada
newData = pca.fit_transform(X_scaled)

df3 = pd.DataFrame(newData)

df3.boxplot()
plt.show()
```

```python
pca = PCA()
df_85 = pca.fit_transform(X_scaled)
varianza_acumulada = pca.explained_variance_ratio_.cumsum()
n_component = (varianza_acumulada >= 0.85).argmax() + 1
n_component
```

```python
pca = PCA(n_components=n_component)
data_85 = pca.fit_transform(X_scaled)
df_85 = pd.DataFrame(data_85)
df_85.boxplot()
plt.show()
```

```python
df_pca = pd.DataFrame(data_85, columns=[f'PC{i+1}' for i in range(n_component)])
df_pca.head()
```
