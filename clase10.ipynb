## Ejercicio 1: 
Calcule el valor de una pdf Gaussiana, N(m,S), en x1 = [0.2, 1.3]<sup>T</sup> y x2 = [2.2,-1.3]<sup>T</sup>, donde

m = [0,1]<sup>T</sup>, S = $\begin{bmatrix} 1 & 0 \\ 0 & 1 \end{bmatrix}$

```python
import numpy as np

def gaussian_pdf(x, m, S):
    """
    Calcula la función de densidad de probabilidad gaussiana (funcion de verosimilitud).
    
    Parametros:
    x: vector (numpy array)
    m: mean vector (numpy array)
    S: matrix de covarianza (numpy array)
    
    Retorna:
    p(x): valor de la función de densidad de probabilidad en x (float)
    """
    l = len(m)
    diff = x - m
    S_inv = np.linalg.inv(S)
    S_det = np.linalg.det(S)
    
    coefficient = 1 / ((2 * np.pi) ** (l / 2) * np.sqrt(S_det))
    exponent = -0.5 * diff.T @ S_inv @ diff # @ es el operador de producto matricial
    
    return coefficient * np.exp(exponent)

# Define the parameters
m = np.array([0, 1])
S = np.array([[1, 0], [0, 1]])

# Define the points
x1 = np.array([0.2, 1.3])
x2 = np.array([2.2, -1.3])

# Calculate the PDF values
pdf_x1 = gaussian_pdf(x1, m, S)
pdf_x2 = gaussian_pdf(x2, m, S)

print(f"p(x1) = {pdf_x1}")
print(f"p(x2) = {pdf_x2}")
```

## Ejercicio 2: 
Considere una tarea de clasificación de dos clases en un espacio de dos-dimensiones, donde los datos en ambas clases, w1, w2, estan distribuidas de acuerdo a las distribuciones Gaussianas N(m1, S1) y N(m2, S2), respectivamente. Sean

m1 = [1,1]<sup>T</sup>, m2 = [3,3]<sup>T</sup> S1 = S2 = $\begin{bmatrix} 1 & 0 \\ 0 & 1 \end{bmatrix}$

Suponga que P(w1) = P(w2) = 1/2, clasificar x = [1.8, 1.8]<sup>T</sup> en w1 o w2

```python
# Define parameters for both classes
m1 = np.array([1, 1])
m2 = np.array([3, 3])
S1 = np.array([[1, 0], [0, 1]])
S2 = np.array([[1, 0], [0, 1]])

# Point to classify
x = np.array([1.8, 1.8])

# Prior probabilities
P_w1 = 0.5
P_w2 = 0.5

# Calculate likelihoods (PDFs)
p_x_w1 = gaussian_pdf(x, m1, S1)
p_x_w2 = gaussian_pdf(x, m2, S2)

# Calculate posterior probabilities (proportional to)
p_w1_x = p_x_w1 * P_w1
p_w2_x = p_x_w2 * P_w2

# Determine the class
class_array = np.array([p_w1_x, p_w2_x])
clase_x = np.argmax(class_array) + 1

print(f"p(x|w1) = {p_x_w1}")
print(f"p(x|w2) = {p_x_w2}")
print(f"P(w1|x) ∝ {p_w1_x}")
print(f"P(w2|x) ∝ {p_w2_x}")

# Classification decision
print(f"\nClasificación: x = {x} pertenece a la clase w{clase_x}")
```
